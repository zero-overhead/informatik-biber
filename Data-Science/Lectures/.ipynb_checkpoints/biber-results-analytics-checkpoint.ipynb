## Vorverarbeitung

```python
df[df.select_dtypes(include="object").columns] = (
    df.select_dtypes(include="object").astype("category")
)
```

```python
df.info()
```

## mit GroupBy und Pivot analysieren

```python
df[["Jahr", "Schuljahre", "Schule", "Punkte"]]\
    .groupby(["Jahr", "Schule", "Schuljahre"], observed=True)\
    .mean()\
    .sort_values(["Jahr", "Punkte"], ascending=[True, False])
```

```python
df.pivot_table(
    columns = ['Schuljahre', 'Schule'],
    index=['Jahr', 'Team'],
    values='Punkte',
    aggfunc='mean',
    observed=True
)
```

## Grafiken erstellen

```python
df.plot(kind="box", column="Punkte", by="Schuljahre", showmeans=True, observed=True)
```

```python
top10 = (
    df.groupby("Aufgabe")["Punkte"]
      .mean()
      .nlargest(10)
      .index
)

df_top10 = df[df["Aufgabe"].isin(top10)].copy()

df_top10["Aufgabe"] = pd.Categorical(
    df_top10["Aufgabe"],
    categories=top10,
    ordered=True
)

df_top10.plot(
    kind="box",
    column="Punkte",
    by="Aufgabe",
    showmeans=True,
    rot=90
)
```

```python
top10 = (
    df.groupby("Aufgabe")["Punkte"]
      .mean()
      .nsmallest(10)
      .index
)

df[df["Aufgabe"].isin(top10)].plot(
    kind="box",
    column="Punkte",
    by="Aufgabe",
    showmeans=True,
    rot=90
)
```
